During a project, we may:

- modify code many times;
- make mistakes and want an older version;
- work from different computers;
- collaborate with colleagues or coders from around the world;
- use different versions of Python packages.

There are tools which help us manage these problems

- **Git** : Tracks changes in files on our computer
- **GitHub** : Stores and shares Git repositories online
- **Virtual environment** :  Keeps python packages isolated for each project

> Git is not the same as GitHub. Git works locally. GitHub is an online hosting service for Git repositories.


# Important Git Termonology

| Term              | Meaning                                       |
| ----------------- | --------------------------------------------- |
| Repository        | A project folder tracked by Git               |
| Working directory | Files currently present in the project folder |
| Staging area      | Files selected for the next commit            |
| Commit            | A saved checkpoint of the project             |
| Branch            | An independent line of development            |
| Remote            | An online copy of the repository              |
| Clone             | Download a repository for the first time      |
| Pull              | Download and combine new changes              |
| Push              | Upload local commits                          |

# Instructions: Create and Use Python Virtual Environment for ML Course

## What is a virtual environment?

A **virtual environment** is a separate Python workspace for one project.

It helps because each project can have its own packages.

Example:

```text
ML_course uses:
numpy
pandas
matplotlib
scikit-learn
jupyter
```

Without virtual environment, packages may mix with other projects and create problems.

---

##  Open Git Bash or Terminal

On Windows, open the folder where your project is located or you want to work.

Example:

```bash
cd ~/Documents/ML_course
```

Check your location:

```bash
pwd
```

---

## Check Python installation

Run:

```bash
python --version
```

or:

```bash
python3 --version
```

One should see something like:

```text
Python 3.12.x
```

---

##  Create virtual environment

Inside your course folder, run:

```bash
python -m venv .venv
```

This creates a folder named:

```text
.venv
```

Your project will look like:

```text
ML_Project/
├── documents/
├── codes/
├── data/
├── examples/
├── .venv/
├── README.md
└── requirements.txt
```

---

## Activate virtual environment

### On Windows Git Bash

Use:

```bash
source .venv/Scripts/activate
```

### On Windows Command Prompt

Use:

```cmd
.venv\Scripts\activate
```

### On Windows PowerShell

Use:

```powershell
.venv\Scripts\Activate.ps1
```

### On Linux / WSL / macOS

Use:

```bash
source .venv/bin/activate
```

After activation, you will see something like:

```text
(.venv)
```

before your terminal line.

Example:

```text
(.venv) user@computer MINGW64 ~/Documents/ML_projet
```

This means the virtual environment is active.

---

## Install packages

After activation, install required packages:

```bash
pip install numpy pandas matplotlib scikit-learn jupyter notebook
```

For plotting and ML, this is enough for starting.

Later, you can add:

```bash
pip install seaborn scipy
```

For deep learning, later add only when needed:

```bash
pip install torch torchvision
```

or:

```bash
pip install tensorflow
```

---

## Create `requirements.txt`

After installing packages, save the package list:

```bash
pip freeze > requirements.txt
```

This file records all packages used in your course.

Students can later install the same packages using:

```bash
pip install -r requirements.txt
```

---

## Start Jupyter Notebook

Make sure virtual environment is active.

Then run:

```bash
jupyter notebook
```

or:

```bash
jupyter lab
```

If JupyterLab is not installed:

```bash
pip install jupyterlab
```

Then:

```bash
jupyter lab
```

---

## Add virtual environment to Jupyter kernel

Install kernel package:

```bash
pip install ipykernel
```

Add this environment to Jupyter:

```bash
python -m ipykernel install --user --name ML_course_env --display-name "Python ML Course"
```

Now when you open Jupyter Notebook, choose kernel:

```text
Python ML Course
```

---

##  Deactivate virtual environment

When work is finished, run:

```bash
deactivate
```

The `(.venv)` mark will disappear.

---

## Add `.venv` to `.gitignore`

Do **not** upload `.venv` to GitLab.

Open `.gitignore` and add:

```text
.venv/
__pycache__/
.ipynb_checkpoints/
*.pyc
.env
```

This keeps your GitLab repository clean.

---

## Daily workflow

Every day before working:

```bash
cd ~/Documents/ML_course
source .venv/Scripts/activate
jupyter notebook
```

After editing notebooks:

```bash
git status
git add .
git commit -m "Updated lecture notebook"
git push
```

After finishing:

```bash
deactivate
```


#  Create and Use Conda Environment for ML Course

## What is a Conda environment?

A **Conda environment** is a separate workspace for one project.

It keeps Python and packages separate from other projects.

For example, your ML project may use:

```
numpy
pandas
matplotlib
scikit-learn
jupyter
notebook
```

Another project may use different versions. Conda helps avoid package conflicts.

---

## Conda vs virtualenv

```
virtualenv / venv = simple Python environment
conda = Python + scientific packages + ML packages
```

For an ML project, Conda is very useful because it handles scientific packages easily.

Recommended:

```
Use Conda for ML course
Use venv for simple Python projects
```

---

## Install Miniconda

Use **Miniconda**, not full Anaconda, because it is smaller and cleaner.

Download Miniconda from the official website and install it.

During installation on Windows, you may see options like:

```
Add Miniconda to PATH
Register Miniconda as default Python
```

For beginners, it is usually safer to use **Anaconda Prompt** after installation.

---

## Open Anaconda Prompt

On Windows:

```
Start Menu → Anaconda Prompt
```

or:

```
Start Menu → Miniconda Prompt
```

You can also use Git Bash later, but first use Anaconda Prompt because Conda works there directly.

---

## Go to your project folder

Example:

```
cd Documents
cd ML_project
```

If your folder is inside OneDrive or another location, go to that path.

Example:

```
cd C:\Users\YourName\Documents\ML_project
```

---

## Create a Conda environment

Create an environment named:

```
ml_project
```

Command:

```
conda create -n ml_project python=3.12
```

Here:

```
-n ml_project    means environment name
python=3.12     means Python version
```

When Conda asks:

```
Proceed ([y]/n)?
```

Type:

```
y
```

and press Enter.

---

## Activate Conda environment

Run:

```
conda activate ml_course
```

After activation, you will see:

```
(ml_project)
```

at the beginning of the terminal line.

Example:

```
(ml_course) C:\Users\YourName\Documents\ML_project>
```

This means your Conda environment is active.

---

## Install basic ML packages

Install important packages:

```
conda install numpy pandas matplotlib scikit-learn jupyter notebook ipykernel
```

When asked:

```
Proceed ([y]/n)?
```

type:

```
y
```

---

## Optional: install from conda-forge

Sometimes packages are newer on `conda-forge`.

You can use:

```
conda install -c conda-forge numpy pandas matplotlib scikit-learn jupyterlab notebook ipykernel
```

For beginners, either normal `conda install` or `conda-forge` is fine.

---

## Install JupyterLab

If you want JupyterLab also:

```
conda install -c conda-forge jupyterlab
```

Then you can run:

```
jupyter lab
```

For simple teaching, `jupyter notebook` is also fine.

---

##  Add Conda environment to Jupyter

This step makes your environment visible inside Jupyter.

Run:

```
python -m ipykernel install --user --name ml_course --display-name "Python ML Course"
```

Now when Jupyter opens, you can select kernel:

```
Python ML Project
```

---

## Start Jupyter Notebook

Make sure the environment is active:

```
(ml_project)
```

Then run:

```
jupyter notebook
```

or:

```
jupyter lab
```

Open your notebook from:

```
ml_project/
```

Example:

```
ml_project/code_1.ipynb
```

---

##  Check Python inside notebook

In a notebook code cell, run:

```
import sys
print(sys.executable)
```

It should show a path related to:

```
ml_course
```

This confirms that the notebook is using the correct Conda environment.

---

## Save environment as `environment.yml`

After installing packages, save the environment:

```
conda env export --from-history > environment.yml
```

This creates a file:

```
environment.yml
```

This file should be uploaded to GitLab.

It tells students which packages are needed.

---

## Example `environment.yml`

A simple `environment.yml` may look like this:

```
name: ml_project
channels:
  - defaults
dependencies:
  - python=3.12
  - numpy
  - pandas
  - matplotlib
  - scikit-learn
  - jupyter
  - notebook
  - ipykernel
```

If using conda-forge:

```
name: ml_project
channels:
  - conda-forge
  - defaults
dependencies:
  - python=3.12
  - numpy
  - pandas
  - matplotlib
  - scikit-learn
  - jupyterlab
  - notebook
  - ipykernel
```

---

##  How you can create the same environment

You should first clone or download your GitLab course folder.

Then they run:

```
cd ML_project
conda env create -f environment.yml
conda activate ml_project
jupyter notebook
```

This creates the same environment on their computer.

---

## Update environment after adding new packages

Suppose later you install a new package:

```
conda install seaborn
```

Then update `environment.yml`:

```
conda env export --from-history > environment.yml
```

Then upload the updated file to GitLab:

```
git add environment.yml
git commit -m "Updated conda environment"
git push
```

---

##  Install package using pip inside Conda environment

Sometimes a package may not be available in Conda.

Then activate the environment:

```bash
conda activate ml_project
```

Install using pip:

```bash
pip install package_name
```

Example:

```bash
pip install kaggle
```

After that, you can record the environment again:

```bash
conda env export > environment.yml
```

For clean project environments, prefer Conda packages first, then pip only when needed.

---

##  Deactivate Conda environment

When work is finished:

```
conda deactivate
```

The environment name disappears from the terminal.

---

## Remove Conda environment if needed

If you make a mistake and want to delete the environment:

```
conda remove -n ml_project --all
```

Then create it again:

```
conda env create -f environment.yml
```


| Feature                  | venv             | conda           |

| Beginner-friendly for ML | Good             | Better          |

| Size                     | Small            | Bigger          |
